Bài 8: Nhập n 
Cho S(k) = 1 + 2 + 3 + … + k 
Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n 

```python
def find_k(n):
    k = 0
    s_k = 0
    while s_k + (k + 1) < n:
        k += 1
        s_k += k
    return k, s_k
try:
    n = int(input("Nhập n: "))
    if n <= 1:
        print("Không có số k dương nào thỏa mãn S(k) < n.")
    else:
        k, tong = find_k(n)
        print(f"Số k lớn nhất là: {k}")
        print(f"Tổng S({k}) = {tong} (nhỏ hơn {n})")
except ValueError:
    print("Vui lòng nhập một số nguyên.")
```

Bài 10: Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1. 
Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập

```python
def solve_min_max():
    numbers = []
    print("Nhập vào các số nguyên (nhập -1 để dừng):")
    while True:
        try:
            n = int(input("Nhập số: "))
            if n == -1:
                break
            numbers.append(n)
        except ValueError:
            print("Vui lòng chỉ nhập số nguyên!")
    if len(numbers) > 0:
        print("--- Kết quả ---")
        print(f"Số lớn nhất là: {max(numbers)}")
        print(f"Số nhỏ nhất là: {min(numbers)}")
    else:
        print("Bạn chưa nhập số nào hợp lệ trước khi kết thúc.")
solve_min_max()
```

```python
max_val = None
min_val = None
while True:
    num = int(input("Nhập một số nguyên (nhập -1 để dừng): "))
    if num == -1:
        break
    if max_val is None:
        max_val = num
        min_val = num
    else:
        if num > max_val:
            max_val = num
        if num < min_val:
            min_val = num
if max_val is not None:
    print(f"Số lớn nhất là: {max_val}")
    print(f"Số nhỏ nhất là: {min_val}")
else:
    print("Bạn chưa nhập số nào ngoài số -1.")
```

Bài 11: Fibonaci1] 
Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ 
bằng tổng hai số trước đó 
Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt 
quá A 

```python
def fibonacci_max():
    A = int(input("Nhập số nguyên dương A: "))
    f1, f2 = 1, 1
    if A < 1:
        print("Không có số Fibonacci thỏa mãn.")
        return
    while True:
        f_next = f1 + f2
        if f_next > A:
            break
        f1 = f2
        f2 = f_next
    print(f"Số Fibonacci lớn nhất không vượt quá {A} là: {f2}")
fibonacci_max()
```

Bài 16: 
Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu ký tự in hoa, bao nhiêu ký tự in 
thường, bao nhiêu ký tự số 

```python
def count_characters(s):
    hoa = 0
    thuong = 0
    so = 0
    for char in s:
        if char.isupper():
            hoa += 1
        elif char.islower():
            thuong += 1
        elif char.isdigit():
            so += 1
            
    return hoa, thuong, so
string = input("Nhập vào một chuỗi: ")
chu_hoa, chu_thuong, chu_so = count_characters(string)
print(f"Số ký tự in hoa: {chu_hoa}")
print(f"Số ký tự in thường: {chu_thuong}")
print(f"Số ký tự số: {chu_so}")
```

```python
s = input("Nhập chuỗi: ")
length = 0
for _ in s:
    length += 1
hoa = 0
thuong = 0
so = 0
for i in range(length):
    ky_tu = s[i]
    if 'A' <= ky_tu <= 'Z':
        hoa += 1
    elif 'a' <= ky_tu <= 'z':
        thuong += 1
    elif '0' <= ky_tu <= '9':
        so += 1
print("Số ký tự in hoa:", hoa)
print("Số ký tự in thường:", thuong)
print("Số ký tự số:", so)
```

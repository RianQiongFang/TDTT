Bài 2: Nhập vào số nguyên dương a, in ra bảng cửu chương của a 

```python
a = int(input())
print(f"Bảng cửu chương {a}:")
for i in range(1, 11):
    ket_qua = a * i
    print(f"{a} x {i:2} = {ket_qua}")
```

Bài 3: Nhập vào số nguyên dương a và b, in toàn bộ ước chung của a và b

```python
import math
a = int(input())
b = int(input())
ucln = math.gcd(a, b)
print(f"Các ước chung của {a} và {b} là:")
for i in range(1, ucln + 1):
    if ucln % i == 0:
        print(i, end=" ")
```

Bài 5: Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 2^k hay không

```python
n = int(input("Nhập số nguyên dương n: "))
if n > 0 and (n & (n - 1)) == 0:
    print(f"{n} là số có dạng 2^k")
else:
    print(f"{n} không phải là số có dạng 2^k")
```

Bài 6: Nhập vào số nguyên dương n, tính tổng các chữ số của n 

```python
n = int(input("Nhập vào số nguyên dương n: "))
tong = 0
temp = n
while temp > 0:
    chu_so = temp % 10 
    tong += chu_so      
    temp //= 10         
print(f"Tổng các chữ số của {n} là: {tong}")
```

Bài 7: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ 

```python
n = int(input())
chan = 0
le = 0
while n > 0:
    chu_so = n % 10
    if chu_so % 2 == 0:
        chan += 1
    else:
        le += 1
    n//= 10
print(f"Số số chẵn là: {chan}")
print(f"Số số lẻ là: {le}") 
```

Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

W6A20:
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về

```python
def tim_k(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1 
    
ds = [10, 20, 30, 40, 50, 20]
k = 20
ket_qua = tim_k(ds, k)
if ket_qua != -1:
    print(f"Giá trị {k} xuất hiện trong danh sách.")
else:
    print(f"Không tìm thấy giá trị {k} trong danh sách.")
```

W6A21: 
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ 
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

W6A26: 
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 
phần tử đầu và cuối 

Cực đại địa phương: phần tử mà giá trị của nó lớn hơn cả phần tử đứng trước và phần tử đứng sau nó

```python
def tim_CDDP(lst):
    tong = 0
    for i in range(1, len(lst) - 1):
        if lst[i] > lst[i - 1] and lst[i] > lst[i + 1]:
            tong += lst[i]
    return tong

lst = [1, 3, 5, 72, 4, 54, 223, 33, 12]
result = tim_CDDP(lst)
print(result)
```

W6A27: 
Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước). 
nếu không có in ra -1 

```python
def tim_so_be(lst, x):
    vi_tri = []
    for i in range(len(lst)):
        if lst[i] < x:
            vi_tri.append(i)
    if len(vi_tri) == 0:
        return -1
    return vi_tri

lst = [10, 5, 20, 15, -4, -100]
result = tim_so_be(lst, 10)
print(result)
    
```

W6A28: 
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm 
thực sự, hay vô danh (không tăng, không giảm)

```python
def kiem_tra_day_so(lst):
    tang = True
    giam = True
    for i in range(len(lst) - 1):
        if lst[i] < lst[i + 1]:
            giam = False
        elif lst[i] > lst[i + 1]:
            tang = False
    if tang:
        return "Dãy tăng"
    elif giam:
        return "Dãy giảm"
    else:
        return "Dãy vô danh"
    
lst1 = [1, 3, 5, 7, 9]
lst2 = [10, 8,6,4, 2, 0]
lst3 = [0, 9, 0 , 3, 0, 7]
print(kiem_tra_day_so(lst1))
print(kiem_tra_day_so(lst2))
print(kiem_tra_day_so(lst3))
```

Cho hai mảng đã sắp xếp (tăng hoặc giảm dần). Hãy trộn hai mảng này thành 1 mảng theo thứ tự sắp xếp mà không được dùng bất cứ 1 thuật toán sắp xếp nào.

```python
lst1 = list(map(int, input().split()))
lst2 = list(map(int, input().split()))

ket_qua = []
i = 0 
j = 0 

while i < len(lst1) and j < len(lst2):
    if lst1[i] < lst2[j]:
        ket_qua.append(lst1[i])
        i += 1
    else:
        ket_qua.append(lst2[j])
        j += 1


while i < len(lst1):
    ket_qua.append(lst1[i])
    i += 1


while j < len(lst2):
    ket_qua.append(lst2[j])
    j += 1

print(*ket_qua)
```

W6A16:
Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n = int(input("Nhập số hàng n: "))
m = int(input("Nhập số cột m: "))
matrix = []
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i+1}: ").split()))
    matrix.append(row)
print("Ma trận đã nhập là:")
for row in matrix:
    for element in row:
       print(f"{element:4d}", end="")
    print()
```

W6A17:
Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và 

```python
n = int(input("Nhập kích thước n của ma trận vuông: "))
matrix = []
print(f"Nhập các phần tử cho ma trận {n}x{n}:")
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i+1}: ").split()))
    matrix.append(row)
print("Các phần tử trên đường chéo chính là:")
for i in range(n):
    print(matrix[i][i], end=" ")
print()
print("Các phần tử trên đường chéo phụ là:")
for i in range(n):
    print(matrix[i][n - 1 - i], end=" ")
```

W6A18:
Nhâ ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k)

```python
m, n, k  = map(int, input("Nhập vào số hàng m, số cột n và số k: ").split())
matrix = []
for i in range(m):
    row = list(map(int, input(f"Nhập hàng {i+1}: ").split()))
    matrix.append(row)
if k < n:
    tong_hang = sum(matrix[k])
    print(f"Tổng hàng {k}: {tong_hang}")
else:
    print(f"Không có hàng {k}")
if k < m:
    tong_cot = 0
    for i in range(n):
        tong_cot += matrix[i][k]
    print(f"Tổng cột {k}: {tong_cot}")
else:
    print(f"Không có cột {k}")
```

W6A19:
Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất

```python
def gia_tri_max(lst):
    max_val = 0
    for num in lst:
        if num > max_val:
            max_val = num
    return max_val

ds = [10, 5, 20, 15, 20]
ket_qua = gia_tri_max(ds)
print(f"Giá trị lớn nhất là: {ket_qua}")
```

W6A20:
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về

```python
def tim_k(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1 
    
ds = [10, 20, 30, 40, 50, 20]
k = 20
ket_qua = tim_k(ds, k)
if ket_qua != -1:
    print(f"Giá trị {k} xuất hiện trong danh sách.")
else:
    print(f"Không tìm thấy giá trị {k} trong danh sách.")
```

Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu 
mỗi phần tử x trong mảng xuất hiện đúng x lần. 
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa 
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp. 

```python
def is_beautiful(arr):
     freq = {}
    for x in arr:
        if x in freq:
            freq[x] += 1
        else:
            freq[x] = 1
    for x in freq:
        if freq[x] != x:
            return False
    return True
def min_deletions_to_beautiful(arr):
    freq = {}
    for x in arr:
        if x in freq:
            freq[x] += 1
        else:
            freq[x] = 1

    deletions = 0
    for x in freq:
        f = freq[x]
        if f == x:
            continue
        elif f > x:
            deletions += f - x
        else:  # f < x
            deletions += f
    return deletions

arr1 = [1,2,2,3,3,3]
print("Mảng:", arr1)
print("Có đẹp không?", is_beautiful(arr1))
print("Số phần tử cần xóa:", min_deletions_to_beautiful(arr1))
```

1. Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số 
nguyên x 

```python
def is_armstrong(x: int) -> bool:
    if x < 0:
      return False
    s = str(x)
    n = len(s)
    total = sum(int(digit) ** n for digit in s)
    return total == x
assert is_armstrong(153) == True
assert is_armstrong(9474) == True
assert is_armstrong(123) == False
assert is_armstrong(0) == True
assert is_armstrong(-153) == False
print("All test cases passed!")
```

2. Viết hàm kiêểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số 
nguyên x 

```python
def is_perfect_square(x: int) -> bool:
    if x < 0:
        return False
    root = int(x**0.5)
    return root * root == x
assert is_perfect_square(16) == True
assert is_perfect_square(15) == False
assert is_perfect_square(0) == True
assert is_perfect_square(-1) == False
assert is_perfect_square(34) == False
print("All test cases passed!")
```

3. VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên. 


```python
def is_palindrome(x: int) -> bool:
    if x < 0 or (x % 10 == 0 and x != 0):
        return False
    reversed_num = 0
    original_x = x
    while x > 0:
        last_digit = x % 10
        reversed_num = reversed_num * 10 + last_digit
        x //= 10
    return original_x == reversed_num
print(is_palindrome(121))  
print(is_palindrome(-121)) 
print(is_palindrome(10))   
print(is_palindrome(12321))
print(is_palindrome(4663))
```

9. Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không. 

 Ba điểm thẳng hàng khi và chỉ khi:$$(x_2 - x_1)(y_3 - y_1) = (x_3 - x_1)(y_2 - y_1)$$

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (y2 - y1) * (x3 - x2) == (y3 - y2) * (x2 - x1)
print(are_collinear(1, 1, 2, 2, 3, 3)) 
print(are_collinear(1, 1, 2, 2, 3, 4)) 
print(are_collinear(0, 0, 0, 5, 0, 10)) 
print(are_collinear(5, 0, 0, 5, 0, 10)) 
print(are_collinear(65, 4, 0, 5, 0, 10)) 
```

10. Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay 
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3), 
(0,3)], p = (2,1) 

```python
def point_in_quadrilateral(quad, p):
    px, py = p
    ket_qua = []
    for i in range(4):
        ax, ay = quad[i]
        bx, by = quad[(i + 1) % 4]
        huong = (bx - ax) * (py - ay) - (by - ay) * (px - ax)
        ket_qua.append(huong)
        la_trong = all(h >= 0 for h in ket_qua)
    la_ngoai = all(h <= 0 for h in ket_qua)
    return la_trong or la_ngoai
quad = [(0,0), (4,0), (4,3), (0,3)]
p = (2,1)
print(point_in_quadrilateral(quad, p))
```

12. Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list) 

```python
def count_non_zero_digits(numbers):
    count = 0
    for n in numbers:
        s = str(abs(n))
        for digit in s:
            if digit != '0':
                count += 1
    return count
input_list = list(map(int, input("Nhập các số nguyên (cách nhau bởi khoảng trắng): ").split()))
result = count_non_zero_digits(input_list)
print(f"Tổng số lượng các chữ số khác 0 là: {result}")
```

13.  Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3). 

```python
import math
def calculate_angle_abc(x1, y1, x2, y2, x3, y3):
    ba_x, ba_y = x1 - x2, y1 - y2
    bc_x, bc_y = x3 - x2, y3 - y2
    dot_product = ba_x * bc_x + ba_y * bc_y
    mag_ba = (ba_x**2 + ba_y**2)**0.5
    mag_bc = (bc_x**2 + bc_y**2)**0.5
    if mag_ba == 0 or mag_bc == 0:
        return 0.0
    cosine_angle = max(-1, min(1, dot_product / (mag_ba * mag_bc)))
    angle_rad = math.acos(cosine_angle)
    angle_deg = math.degrees(angle_rad)
    return angle_deg
print(calculate_angle_abc(0, 1, 0, 0, 1, 0))
print(calculate_angle_abc(0, 1, 0, 4, 1, 0))
print(calculate_angle_abc(0, 1, 3, 0, 1, 0))
print(calculate_angle_abc(0, 1, 0, 0, 1, 2))
print(calculate_angle_abc(1, 1, 0, 0, 1, 0))
```

15. Viết hàm tính hệ số nhị thức C(n, k). 

Công thức đệ quy: $C(n, k) = C(n-1, k-1) + C(n-1, k)$

```python
def combinations_recursive(n, k):
    if k == 0 or k == n:
        return 1
    if k < 0 or k > n:
        return 0
    return combinations_recursive(n - 1, k - 1) + combinations_recursive(n - 1, k)
print(combinations_recursive(5, 2))
print(combinations_recursive(4, 2))
print(combinations_recursive(6, 9))
print(combinations_recursive(3, 7))
print(combinations_recursive(1, 8))
```

18. Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa, 
giá km, phí chờ, phụ phí ban đêm và giảm giá. 

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4:
        opening_fare = 12000  
        rate_to_25 = 14500   
        rate_from_26 = 12000  
        wait_rate = 1000     
    else: 
        opening_fare = 15000
        rate_to_25 = 17000
        rate_from_26 = 14500
        wait_rate = 1500
    if km <= 1:
        total_km_fare = km * opening_fare
    elif km <= 25:
        total_km_fare = opening_fare + (km - 1) * rate_to_25
    else:
        total_km_fare = opening_fare + (24 * rate_to_25) + (km - 25) * rate_from_26
    chargeable_wait = max(0, wait_minutes - 5)
    total_wait_fare = chargeable_wait * wait_rate
    
    total_fare = total_km_fare + total_wait_fare
    if total_fare > 500000:
        total_fare *= 0.9
    if start_hour >= 22 or start_hour < 5:
        total_fare *= 1.2

    return round(total_fare)

fare = smart_taxi_fare(car_type=4, start_hour=23, km=30, wait_minutes=10)
print(f"Tổng tiền taxi: {fare:,} VNĐ")
```

19.   Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau 

Nếu n ≤ 0 trả về 0. 

```python
def sum_1_to_n(n):
    if n <= 0:
        return 0
    tong = 0
    for x in range(1, n + 1):
        s = str(x)
        if len(s) == len(set(s)):
            tong += x
    return tong
print(sum_1_to_n(15))  
print(sum_1_to_n(25))
print(sum_1_to_n(35))
print(sum_1_to_n(45))
print(sum_1_to_n(55))
```

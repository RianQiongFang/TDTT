Viết chương trình Python sử dụng 1 trong 2 vòng lặp for hoặc while để tính tổng các số nguyên từ 1 đến n (n<=1000).

```python
n = int(input())
tong = 0
for i in range (1, n + 1):
    tong += i
print(tong)
```

```python
n = int(input("Nhập số nguyên n (1 <= n <= 1000): "))
if 1 <= n <= 1000:
    tong = 0
    i = 1
    while i <= n:
        tong += i  
        i += 1     
    print(f"Tổng các số từ 1 đến {n} là: {tong}")
else:
    print("Vui lòng nhập n trong khoảng từ 1 đến 1000.")
```

VIết chương trình nhập vào một số tới khi được một số nguyên dương thì thôi. 
Kiểm tra số này có phải là số nguyên tố hay không?

```python
while True:
    n = int(input())
    if n<0:
        print("Nhập lại:")
    else:
        break
def la_so_nguyen_to(n):
    if n<2:
        return False
    for i in range (2, int(n**0.5) +1):
        if n % i ==0:
            return False
        return True
if la_so_nguyen_to(n): 
    print(f"{n} là số nguyên tố") 
else:
    print(f"{n} không phải là số nguyên tố")  
```

Nhập hai số nguyên dương a, b (a ≤ b). Hãy tính tổng các số nguyên tố trong đoạn 
[a, b]. Mỗi dòng là một bộ test (a b).

```python
def la_so_nguyen_to(n):
    if n<2:
        return False
    for i in range (2, int(n**0.5) +1):
        if n % i ==0:
         return False
    return True
while True:
      a, b = map(int, input().split()) 
      tong = 0
      for i in range (a, b+1):
        if la_so_nguyen_to(i):
          tong += i 
      print(f"Tổng các số nguyên tố trong đoạn [{a}, {b}] là: {tong}")
      break
```

Hãy viết một chương trình nhận vào một số nguyên dương n và một dãy gồm n số 
nguyên. Nếu n số nguyên đó có số 42 thì in ra màn hình dòng chữ "I′ve found the 
meaning of life!", ngược lại in ra dòng chữ "It′s a joke!".

```python
n = int(input())
numbers = list(map(int, input().split()))
numbers = numbers[:n]
if 42 in numbers:
    print("I've found the meaning of life!")
else:
    print("It's a joke!")
```

### Bài 1: Tìm số lớn nhất của 1 list

```python
def so_lon_nhat(lst: list) -> int:
    if not lst:
        return None
    max_val = lst[0]
    for x in lst:
        if x > max_val:
            max_val = x
    return max_val
lst = list(map(int, input().split()))
print(so_lon_nhat(lst))
```

### Bài 2: Tìm số nhỏ nhất của 1 list

```python
def so_be_nhat(lst: list) -> int:
    if not lst:
        return None
    min_val = lst[0]
    for x in lst:
        if x < min_val:
            min_val = x
    return min_val
lst = list(map(int, input().split()))
print(so_be_nhat(lst))
```

### Bài 2b: Tìm số nhỏ nhất của 1 list và vị trí của nó, nếu có nhiều số nhỏ nhất bằng nhau thì lấy chỉ số đầu tiên (nhỏ nhất)
10 0 0 12 34 5 6 7  8 89 
0 1

```python
def tim_min_va_vi_tri(lst):
    if not lst:
        return None
    min_val = lst[0]
    min_index = 0
    for i, x in enumerate(lst):
        if x < min_val:
            min_val = x
            min_index = i
    return min_val, min_index
numbers = [10, 5, 2, 8, 2, 7]
gia_tri, vi_tri = tim_min_va_vi_tri(numbers)
print(f"Số nhỏ nhất là: {gia_tri}")
print(f"Vị trí đầu tiên xuất hiện: {vi_tri}")
```

### Bài 3: Tìm giá trị trung bình của 1 list

```python
def tong(lst: list) -> int:
    tong = 0
    for x in lst:
        tong += x
        trung_binh = tong / len(lst)
    return tong, trung_binh
lst = list(map(int, input().split()))
print(tong(lst))
```

Bài 3b: Tìm giá trị trung bình của các số nguyên dương trong 1 list

```python
lst = list(map(int, input().split()))
tong = 0
dem = 0
for x in lst:
    if x > 0:
        tong += x
        dem += 1
        trung_binh = tong / dem if dem > 0 else 0
print(trung_binh)
```

### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang
9 8 7 6 5 4 0: Giam
1 2 3 7 1 0 2: Vo danh
1 2 2 4 5 7: Vo danh

Vô danh: Khi có ít nhất một cặp không tăng VÀ ít nhất một cặp không giảm, hoặc có các cặp bằng nhau ($lst[i] == lst[i+1]$).

```python
def kiem_tra_day_so(nums):
    if len(nums) < 2:
        return "Dãy quá ngắn để kết luận (mặc định là tăng/giảm)"
    is_increasing = True
    is_decreasing = True
    for i in range(len(nums) - 1):
        if nums[i] < nums[i+1]:
            is_decreasing = False  
        elif nums[i] > nums[i+1]:
            is_increasing = False  
        else:
            is_increasing = False
            is_decreasing = False
    if is_increasing:
        return "Dãy tăng"
    elif is_decreasing:
        return "Dãy giảm"
    else:
        return "Dãy vô danh"
test_lists = [
    [1, 3, 5, 7, 9],    
    [10, 8, 4, 2, 1],  
    [1, 5, 2, 8, 3],    
    [1, 2, 2, 3]        
]

for lst in test_lists:
    print(f"Dãy {lst}: {kiem_tra_day_so(lst)}")
```

### Bài 8: Kiểm tra một list số nguyên có phải là ziczac khong
1 3 2 5 2 8 7 9 2 100: ziczac
34 2 56 7 90 20 30 7 120: ziczac

Định nghĩa Ziczac: Là dãy mà các số thay đổi chiều liên tục: Tăng -> Giảm -> Tăng hoặc Giảm -> Tăng -> Giảm.

```python
lst = list(map(int, input().split()))
check = False
for i in range(len(lst)-1):
    if lst[i-1] < lst[i] > lst[i+1] or lst[i-1] > lst[i] < lst[i+1]:
        check = True
        break
print("Zigzag") if check else print("Not Zigzag")
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước. Nếu không có k trong ds thì in ra "-1".
VD: k = 3, [1,3,4,4,3,3,4,3,6,1] => k xuất hiện tại các vị trí (trả về 1 list) [1, 4, 5, 7]
k = 10, [1,3,4,4,3,3,4,3,6,1] => k không xuất hiện trong ds => trả về "-1"

```python
k = int(input())
lst = list(map(int, input().split()))
for i, v in enumerate(lst):
    if v == k:
        print(i)
        break
    else:
        print(-1)
    
```

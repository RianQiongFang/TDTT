BÀI 2: Yêu cầu: Cho số nguyên x, trả về True nếu tồn tại số nguyên n ≥ 1 sao cho x = 1 + 2 + ... + n, ngược lại trả về False. 
Khuyến khích dùng vòng lặp đơn giản để cộng dồn tổng. 

```python
x = int(input())
tong = 0
n = 1
ket_qua = False
if x >= 1:
    while tong < x:
        tong = tong + n
        if tong == x:
            ket_qua = True
            break
        n = n + 1
print(f"Số n cần tìm để tổng 1 + 2 + ... + n bằng {x} là: {n}")
print(ket_qua)
```

BÀI 3: 
Yêu cầu: Cho tọa độ ba điểm A(x1,y1), B(x2,y2), C(x3,y3), trả về True nếu chúng tạo thành một tam giác, ngược lại trả về 
False. 
Quy tắc: Chúng tạo thành tam giác nếu không thẳng hàng (diện tích có hướng khác 0).

Ba điểm $A(x_1, y_1)$, $B(x_2, y_2)$, và $C(x_3, y_3)$ sẽ thẳng hàng nếu diện tích tam giác tạo bởi chúng bằng 0. Công thức tính giá trị đại diện cho diện tích (không chia 2) là:$$S = x_1(y_2 - y_3) + x_2(y_3 - y_1) + x_3(y_1 - y_2)$$Nếu $S \neq 0$, ba điểm tạo thành một tam giác.

```python
x1 = float(input("x1 = "))
y1 = float(input("y1 = "))
x2 = float(input("x2 = "))
y2 = float(input("y2 = "))
x3 = float(input("x3 = "))
y3 = float(input("y3 = "))
s = x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)
if s != 0:
    print(True)
else:
    print(False)
```

sách được sắp xếp tăng dần.  
Quy tắc: - Nếu a == 0 và b != 0: trả về nghiệm duy nhất [-c / b]. - Nếu a == 0 và b == 0: trả về [] (không có nghiệm hữu hạn duy nhất nếu c == 0, hoặc vô nghiệm nếu c != 0). - Với a != 0, dùng biệt thức d = bb - 4ac: - d < 0 → [] - d == 0 → [ -b / (2a) ] - d > 0 → hai nghiệm [r1, r2] với r1 <= r2. 

```python
a, b, c = map(float, input().split())
if a == 0:
    if b == 0:
        if c != 0:
           print("Phương trình vô nghiệm")
        else:
            print("Phương trình vô số nghiệm")
    else: 
        x = -c/b
        print("Phương trình có nghiệm duy nhất x =", x)
else: 
    delta = b**2 - 4*a*c
    if delta < 0:
        print("Phương trình vô nghiệm")
    elif delta == 0: 
        x = -b/(2*a)
        print("Phương trình có nghiệm kép x =", x)
    else:
        x1 = (-b + delta**0.5) / (2*a)
        x2 = (-b - delta**0.5) / (2*a)
        print("Phương trình có hai nghiệm phân biệt x1 =", x1, "và x2 =", x2)  
```

BÀI 6: Yêu cầu: Cho ba số a, b, c, trả về giá trị ở giữa khi sắp xếp ba số theo thứ tự tăng dần (giá trị lớn thứ hai).

Định nghĩa này vẫn đúng khi có phần tử trùng nhau (ví dụ: 2, 2, 5 → giá trị giữa là 2). 

```python
a, b, c = map(float, input().split())
if (b <= a <= c) or (c <= a <= b):
    gia_tri_giua = a
elif (a <= b <= c) or (c <= b <= a):
    gia_tri_giua = b
else:
    gia_tri_giua = c
print(gia_tri_giua)
```

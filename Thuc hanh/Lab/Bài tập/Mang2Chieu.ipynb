Bài 3: Trả về cột (chỉ số bắt đầu từ 0) có tổng cột lớn nhất trong ma trận n x m. 
Viết hàm column_with_max_sum(mat) trả về chỉ số bắt đầu từ 0 của cột có tổng lớn nhất. 
Nếu có nhiều cột cùng tổng lớn nhất, trả về cột có chỉ số nhỏ nhất. - Dùng vòng lặp để tính tổng từng cột. - Không dùng tuple hoặc dictionary. 
n = 2
m = 3
mat = [[1, 3, 5], 
       [6, 2, -9]]


```python
def column_with_max_sum(matrix):
    max_sum = float('-inf')
    max_column = 0
    for j in range(m):
        column_sum = sum(matrix[i][j] for i in range(n))
        if column_sum > max_sum:
            max_sum = column_sum
            max_column = j
    return max_column

n = int(input("Nhập số hàng n: "))
m = int(input("Nhập số cột m: "))
matrix = []
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i+1}: ").split()))
    matrix.append(row)
print(column_with_max_sum(matrix))
```

Bài 5: 
Cho ma trận vuông cấp n. In ra đường chéo chính và đường chéo phụ của ma trận


```python
n = int(input("Nhập số hàng n: "))
m = int(input("Nhập số cột m: "))
matrix = []
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i+1}: ").split()))
    matrix.append(row)
print("Các phần tử trên đường chéo chính là:")
for i in range(n):
    print(matrix[i][i], end=" ")
print()
print("Các phần tử trên đường chéo phụ là:")
for i in range(n):
    print(matrix[i][n - 1 - i], end=" ")
```

Bài 6: Cho ma trận nxm số nguyên, thêm vào một cột tại vị trí k.

```python
n = int(input("Nhập số hàng n: "))
m = int(input("Nhập số cột m: "))
k = int(input("Nhập cột k: "))
cot_moi = list(map(int, input("Nhập cột mới: ").split()))
matrix = []
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i+1}: ").split()))
    matrix.append(row)
if k < m:
    for i in range(len(matrix)):
        matrix[i].insert(k, cot_moi[i])
    print(f"Ma trận sau chèn cột mới: ") 
    for row in matrix:
        print(*row)
else:
    print("Cột không tồn tại.")
```

Bài 7: Cho ma trận số nguyên nxm, xóa đi hàng k (0 <= k < n).

```python
n = int(input("Nhập số hàng n: "))
m = int(input("Nhập số cột m: "))
k = int(input("Nhập cột k: "))
matrix = []
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i+1}: ").split()))
    matrix.append(row)
if 0 <= k< len(matrix):
    matrix.pop(k)
print(matrix)
    
```

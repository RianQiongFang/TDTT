```python
try: 
    #Khối lệnh có thể xảy ra lỗi
    a, b = map(int, input().split())
    print(a/b)
except ZeroDivisionError as v:
    print("Lỗi chia cho KHÔNG")
except ValueError as s:
    print("Dữ liệu nhập sai rồi")
except Exception:
    print("Gặp lỗi quay lại hỏi người viết")
else:
    print("Chạy bình thường")
finally:
    print("All passed")
```

```python
def solve(a, b):
    try: 
        #Khối lệnh có thể xảy ra lỗi
        a, b = map(int, input().split())
        print(a/b)
        return 100
    except ZeroDivisionError as v:
        print("Lỗi chia cho KHÔNG")
    except ValueError as s:
        print("Dữ liệu nhập sai rồi")
    except Exception:
        print("Gặp lỗi quay lại hỏi người viết")
    else:
        print("Chạy bình thường")
    finally:
        print("All passed")
print(solve(a, b))
```

W6A1: Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ 
nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả. 
Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3] 

```python
lst = list(map(int, input().split()))
day_moi = []
for i in lst:
    if i not in day_moi:
        day_moi.append(i)
print(day_moi)
```

W6A2: Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu tới vị 
trí i), giữ nguyên thứ tự 

```python
lst = list(map(int, input().split()))
ket_qua = []
tong_hien_tai = 0
for i in lst:
    tong_hien_tai += i
    ket_qua.append(tong_hien_tai)
print(ket_qua)
```

W6A12: [Tìm phần tử xuất hiện nhiều nhất] 
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list. 
Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất 
trong số đó. 

```python
lst = list(map(int, input().split()))
so_dem_nhieu_nhat = lst[0]
so_lan_xuat_hien = 1
so_lan_xuat_hien_max = 1
for i in lst:
    for j in lst:
        if i == j:
            so_lan_xuat_hien += 1
    if so_lan_xuat_hien > so_lan_xuat_hien_max:
        so_lan_xuat_hien_max = so_lan_xuat_hien
        so_dem_nhieu_nhat = i
    elif so_lan_xuat_hien == so_lan_xuat_hien_max:
        if i < so_dem_nhieu_nhat:
            so_dem_nhieu_nhat = i
print(so_dem_nhieu_nhat)
```

W6A0: Cho 1 danh sách số nguyên các giá trị. Liệt kê số lần xuất hiện của các số trong danh sách

```python
lst = list(map(int, input().split()))
ket_qua = []
for x in lst:
    if x not in ket_qua:
        dem = 0
        for y in lst:
            if x == y:
                dem += 1
        print(f"{x}: {dem} lần")
        ket_qua.append(x)
```

1. BUBBLE SORT: Đổi chỗ hai số liên tiếp

```python
lst = [64, 34, 25, 12, 22, 11, 90]
def bubble_sort(lst):
    print("Original list: ", lst)
    n = len(lst)
    result = False
    for i in range(n):
        for j in range(0, n - i - 1):
            if lst[j] > lst[j + 1]:
                lst[j], lst[j + 1] = lst[j + 1], lst[j]
                result = True
        print(lst)
        if not result:
            break

bubble_sort(lst)
```

2. SELECTION SORT: Tìm phần tử nhỏ nhất rồi đẩy lên đầu

3. INSERTION SORT: So sánh ngược về trước rồi "chèn" nó vào đúng vị trí của nó trong đoạn đã sắp xếp

```python
lst = [64, 34, 25, 12, 22, 11, 90]
def insertion_sort(lst):
    n = len(lst)
    for i in range(1, n):
        key = lst[i]
        j = i - 1
        while j >= 0 and key < lst[j]:
            lst[j + 1] = lst[j]
            j -= 1
        lst[j + 1] = key
        print(lst)
insertion_sort(lst)
        
```

* Dùng hàm đệ qui để in ra biểu diễn nhị phân của 1 số nguyên

```python
def binary_search(x: int):
    if x <= 1:
        print(x)
        return x
    binary_search(x // 2)
    print(x % 2, end ='') 
binary_search(10)

```

* Tìm số fibonaci thứ n: 0, 1, 1, 2, 3, 5, 8,....

```python
def fibonacci(n: int):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
n = int(input())
print(fibonacci(n - 1))
```

```python
lst1 = [1, 2, -1, 5, 7, 9, 11, 42]
lst2 = [-3, 5, 67, 322]
lst = lst1 + lst2
def merge(lst, st, mid, en):
    i = st
    j = mid 
    res = []
    while i < mid and j < en:
        if lst[i] <= lst[j]:
            res.append(lst[i])
            i += 1
        else:
            res.append(lst[j])
            j += 1
    while i < mid:
        res.append(lst[i])
        i += 1
    while j < en:
        res.append(lst[j])
        j += 1
    lst[st:en] = res

def merge_sort(lst, st, en):
    if en - st  <= 1:
        return
    mid = (st + en) // 2
    merge_sort(lst, st, mid)
    merge_sort(lst, mid, en)
    merge(lst, st, mid, en)
merge_sort(lst, 0, len(lst))
print(lst)
    
    
```

W6A1:
Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng
giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.
Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3]


```python
lst = list(map(int, input().split()))
ket_qua = []
for i in range(len(lst)):
    if lst[i] not in ket_qua:
        ket_qua.append(lst[i])
print(ket_qua)
   
          
```

Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu tới vị trí i), giữ nguyên thứ tự

W6A4:
Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp lại, thêm vào list của key đó


```python
data = input().split()
dict = {}
for item in data:
    key, value = item.split(':')
    if key not in dict:
        dict[key] = int(value)
    else:
        dict[key].append(int(value))
print(dict)
```

W6A5
Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ...,
'zeros': ...} (đếm số >0, <0, =0)


```python
nums = list(map(int, input().split()))
result = {
    'positives': 0,
    'negatives': 0,
    'zeros': 0
}
for n in nums:
    if n > 0:
        result['positives'] += 1
    elif n < 0:
        result['negatives'] += 1
    else:
        result['zeros'] += 1

print(result)
```

W6A8: 
Viết chương trình nhập vào một list các chuỗi, sau đó đếm số lần xuất hiện
của từng chuỗi và lưu vào dictionary.


```python
data = input().split()
dem_chuoi = {}
for chuoi in data:
    if chuoi not in dem_chuoi:
        dem_chuoi[chuoi] = 1
    else:
        dem_chuoi[chuoi] += 1
print(dem_chuoi)

```

W6A9: Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một
dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ
được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới
theo thứ tự từ điển tăng dần của key. 

```python
data1 = input().split()
data2 = input().split()
dict1 = {}
dict2 = {}
for item in data1:
    key, value = item.split(':')
    dict1[key] = value
for item in data2:
    key, value = item.split(':')
    dict2[key] = value
ket_qua = {}
for k in dict1:
    ket_qua[k] = dict1[k]
for k in dict2:
    if k in ket_qua:
        ket_qua[k] = ket_qua[k] + dict2[k]
    else:
        ket_qua[k] = dict2[k]
key = list(ket_qua.keys())
n = len(key)
for i in range(n):
    for j in range(i + 1, n):
        if key[i] > key[j]:
            tam = key[i]
            key[i] = key[j]
            key[j] = tam
dict_cuoi_cung = {}
for k in key:
    dict_cuoi_cung[k] = ket_qua[k]
print(dict_cuoi_cung)
```

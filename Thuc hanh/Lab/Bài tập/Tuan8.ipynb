[Cylinder]
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao
(kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện
tích toàn phần) và thể tích của hình trụ
Đầu vào
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và
chiều cao của hình trụ, cách nhau bởi một dấu cách.
Đầu ra
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử
dụng lại những hàm vừa viết, π=3.14).
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy). 



```python
class Cylinder:
    def __init__(self, rr = 0, hh = 0):
        self.r = rr
        self.h = hh
    def get_surface_area(self):
        return 2 * 3.14 * self.r * self.h + 2 * 3.14 * self.r * self.r
    def get_volume(self):
        return 3.14 * self.r * self.r * self.h
    def __str__(self):
        return f"({self.r}, {self.h})"
    def __repr__(self):
        return f"({self.r}, {self.h})"
    
r, h = map(float, input().split())
my_cylinder = Cylinder(r, h)
print(my_cylinder)
```

[Date]
Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương
thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không
và phương thức tính ngày kế tiếp của ngày tháng năm nào đó
nếu nó là hợp lệ
Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn
dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in
ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó.
Đầu vào
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng
dd/mm/yyyy
.
Đầu ra
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng
dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID 

```python
class Date:
    def __init__(self, dd = 0, mm = 0, yy = 0):
        self.d = dd
        self.m = mm
        self.y = yy
    def __str__(self):
        return f"{self.d:02d}/{self.m:02d}/{self.y:04d}"  
    def __check__(self) -> bool:
        max_d = -1
        match self.m:
            case 1|3|5|7|8|10|12:
                max_d = 31
            case 4|6|9|11:
                max_d = 30
            case 2:
                if (self.y % 4 == 0 and self.y % 100 != 0) or (self.y % 400 == 0):
                    max_d = 29
                else:
                    max_d = 28
        return (1 <= self.d <= max_d and self.y >= 0)  
    def getNextDay(self) -> Date:
        max_d = -1
        match self.m:
            case 1|3|5|7|8|10|12: max_d = 31
            case 4|6|9|11: max_d = 30
            case 2:
                if (self.y % 4 == 0 and self.y % 100 != 0) or (self.y % 400 == 0):
                    max_d = 29
                else:
                    max_d = 28
        next_d = self.d + 1
        next_m = self.m
        next_y = self.y
        if next_d > max_d:
            next_d = 1
            next_m += 1
            if next_m > 12:
                next_m = 1
                next_y += 1
        return Date(next_d, next_m, next_y)

dd, mm, yy = map(int, input().split())
my_date = Date(dd, mm, yy)
print(my_date.getNextDay())
```

thức kiểm tra xem có phải gốc tọa dộ không, có nằm trên trục
tung không, có nằm trên trục hoành không, tính khoảng cách từ
điểm đó tới một điểm khác
Đầu vào:
Nhập vào tọa độ một điểm M bất kỳ
Đầu ra:
Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục
tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ
là bao nhiêu (lấy 2 chữ số phần thập phân) 


```python
class Point:
    def __init__(self, x = 0, y = 0):
        self.x = x
        self.y = y
    def is_origin(self) -> bool:
        return self.x == 0 and self.y == 0
    def is_on_x_axis(self) -> bool:
        return self.y == 0 and self.x != 0
    def is_on_y_axis(self) -> bool:
        return self.x == 0 and self.y != 0
    def getDistance(self, other) -> float:
        return ((self.x - other.x) ** 2 + (self.y - other.y) ** 2) ** 0.5

data = input().split()
x = float(data[0])
y = float(data[1])
M = Point(x, y)
O = Point(0, 0)
if M.is_origin():
    print("M la goc toa do")
elif M.is_on_y_axis():
    print("M nam tren truc tung")
elif M.is_on_x_axis():
    print("M nam tren truc hoanh")
else:
    print("M khong nam tren truc nao dac biet")
distance = M.getDistance(O)
print(f"{distance:.2f}")
```

Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản
phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi
sản phẩm là một từ điển có name (tên) và price(giá).
Với các phương thức sau:
1. Khởi tạo với danh sách rỗng
2. Thêm vào giỏ hàng một sản phẩm mới
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)

muốn thoát khỏi chương trình không?" 

```python
class ShoppingCart:
    def __init__(self):
        self.items = []
    def add_item(self, name, price):
        item = {"name": name, "price": price}
        self.items.append(item)
        print(f"Đã thêm {name} với giá {price} vào giỏ hàng.") 
    def remove_item(self, name):
        found = False
        for item in self.items:
            if item["name"] == name:
                self.items.remove(item)
                print(f"Đã xóa {name} khỏi giỏ hàng.") 
                found = True
                break
        if not found:
            print(f"Không tìm thấy {name} trong giỏ hàng.")
    def is_empty(self):
        return len(self.items) == 0
    def get_total_price(self):
        total = 0
        for product in self.items:
            total += product["price"]
        return total
    def display_cart(self):
        if self.is_empty():
            print("Giỏ hàng hiện đang trống.")
        else:
            print("--- DANH SÁCH SẢN PHẨM TRONG GIỎ ---")
            for idx, product in enumerate(self.items, start=1):
                print(f"{idx}. Tên: {product['name']} - Giá: {product['price']:,}đ")
            print(f"-> Tổng số lượng: {len(self.items)} sản phẩm.")
    def clear_cart(self):
        self.items.clear()
        print("Đã xóa toàn bộ sản phẩm trong giỏ hàng.")
    def main():
        my_cart = ShoppingCart()
        while True:
            print("       HỆ THỐNG QUẢN LÝ GIỎ HÀNG       ")
            print("1. Khởi tạo lại giỏ hàng mới (Xóa cũ)")
            print("2. Thêm sản phẩm mới")
            print("3. Xóa sản phẩm theo tên")
            print("4. Kiểm tra giỏ hàng rỗng")
            print("5. Tính tổng tiền giỏ hàng")
            print("6. Hiển thị toàn bộ giỏ hàng")
            print("7. Xóa toàn bộ giỏ hàng")
            choice = input("Mời bạn nhập lựa chọn (1-7): ")
            if choice == '1':
                confirm = input("Bạn có chắc muốn hủy giỏ hiện tại để tạo giỏ mới? (y/n): ")
                if confirm.lower() == 'y':
                    my_cart = ShoppingCart() 
                else:
                    print("Đã hủy thao tác khởi tạo lại.")
            elif choice == '2':
                name = input("Nhập tên sản phẩm: ").strip()
                while True:
                    try:
                        price = float(input("Nhập giá sản phẩm: "))
                        if price < 0:
                            print("Giá tiền không thể âm. Vui lòng nhập lại.")
                            continue
                        break
                    except ValueError:
                        print("Giá tiền phải là một con số. Vui lòng nhập lại.")
                my_cart.add_item(name, price)
            elif choice == '3':
                name_to_remove = input("Nhập chính xác tên sản phẩm cần xóa: ").strip()
                my_cart.remove_item(name_to_remove)
            elif choice == '4':
                if my_cart.is_empty():
                    print("Giỏ hàng đang RỖNG.")
                else:
                    print(f"Giỏ hàng KHÔNG rỗng (Đang có {len(my_cart.cart)} sản phẩm).")
            elif choice == '5':
                total = my_cart.get_total_price()
                print(f"Tổng tiền các sản phẩm trong giỏ hàng là: {total:,}đ")
            elif choice == '6':
                my_cart.display_cart()
            elif choice == '7':
                confirm = input("Bạn có chắc chắn muốn xóa hết giỏ hàng không? (y/n): ")
                if confirm.lower() == 'y':
                    my_cart.clear_cart()
                else:
                    print("Đã hủy thao tác xóa toàn bộ giỏ.")
            else:
                print("Lựa chọn không hợp lệ! Vui lòng nhập số từ 1 đến 7.")
                continue
            exit_prompt = input("\nCó muốn thoát khỏi chương trình không? (Gõ 'yes' để thoát): ").strip()
            if exit_prompt.lower() == "yes":
                print("Cảm ơn bạn đã sử dụng dịch vụ giỏ hàng!")
                break
main()
```

perimeter() trả về chu vi.
scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h),
không trả về gì.
Input
w h k (số thực dương)
Output
area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân).

```python
class Rectangle:
    def __init__(self, w = 0, h = 0):
        self.width = w
        self.height = h
    def get_area(self) -> float:
        return self.width * self.height
    def get_perimeter(self) -> float:
        return 2 * (self.width + self.height)
    def get_scale(self, k: float) -> None:
        self.width += k
        self.height *= k
if __name__ == "__main__":
    w, h, k = map(float, input().split())
    rect = Rectangle(w, h)
    rect.get_scale(k)
    print(f"Diện tích hình chữ nhật sau khi scale: {rect.get_area():.2f} và chu vi hình chữ nhật sau khi scale: {rect.get_perimeter():.2f}")
```

Yêu cầu
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:
Rút gọn phân số trong __init__ (quy ước mẫu > 0).
Các phương thức add/sub/mul/div trả về phân số đã rút gọn.
__str__() in dạng a/b.
Input
a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/})
Output
Phân số kết quả đã rút gọn

```python
from math import gcd
class Fraction:
    def __init__(self, num = 0, den = 1):
        if den == 0:
            raise ValueError("Mẫu số phải khác 0.")
        if den < 0:
            num = -num
            den = -den
        current_gcd = gcd(num, den)
        self.num = num // current_gcd
        self.den = den // current_gcd
    def add(self, other: 'Fraction') -> 'Fraction':
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    def sub(self, other: 'Fraction') -> 'Fraction':
        new_num = self.num * other.den - other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    def mul(self, other: 'Fraction') -> 'Fraction':
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)
    def div(self, other: 'Fraction') -> 'Fraction':
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num, new_den)
    def __str__(self) -> str:
        return f"{self.num}/{self.den}" 
if __name__ == "__main__":
    line = input().split()
    a = int(line[0])
    b = int(line[1])
    op = line[2]
    c = int(line[3])
    d = int(line[4])
    f1 = Fraction(a, b)
    f2 = Fraction(c, d)
    if op == '+':
        result = f1.add(f2)
    elif op == '-':
        result = f1.sub(f2)
    elif op == '*':
        result = f1.mul(f2)
    elif op == '/':
        result = f1.div(f2)
    print(result)
```

Lớp BankAccount(owner, balance) với:
deposit(x), withdraw(x); nếu rút quá số dư thì bỏ qua giao dịch.
__str__() in owner: balance.
Input
Dòng 1: owner initial_balance
Dòng 2: n (số lệnh)
Tiếp theo n dòng, mỗi dòng: DEPOSIT x hoặc WITHDRAW x.
Output
Số dư cuối cùng (in số thực, 2 chữ số)

```python
class BankAccount:
    def __init__(self, owner, balance = 0):
        self.owner = owner
        self.balance = balance
    def deposit(self, x):
        self.balance += x
        return self.balance
    def withdraw(self, x):
        if x > self.balance:
            return "Số dư không đủ để rút." 
        elif x <= self.balance:
            self.balance -= x
            return self.balance
    def __str__(self):
        return f"Account owner: {self.owner}, Balance: {self.balance}"
if __name__ == "__main__":
    owner = input("Nhập tên chủ tài khoản: ") 
    initial_balance = float(input("Nhập số dư tài khoản ban đầu: ")) 
    account = BankAccount(owner, initial_balance)
    while True:
        print("__HỆ THỐNG QUẢN LÝ TÀI KHOẢN NGÂN HÀNG__") 
        print("1. Nạp tiền") 
        print("2. Rút tiền") 
        print("3. Kiểm tra số dư") 
        print("4. Thoát") 
        choice = input("Chọn thao tác (1-4): ") 
        if choice == '1':
            amount = float(input("Nhập số tiền muốn nạp: ")) 
            new_balance = account.deposit(amount)
            print(f"Số dư sau khi nạp: {new_balance}")
        elif choice == '2':
            amount = float(input("Nhập số tiền muốn rút: ")) 
            result = account.withdraw(amount)
            if isinstance(result, str):
                print(result)
            else:
                print(f"Số dư sau khi rút: {result}")
        elif choice == '3':
            print(account)
        elif choice == '4':
            print("Cảm ơn bạn đã sử dụng dịch vụ!")
            break
        else:
            print("Lựa chọn không hợp lệ. Vui lòng chọn lại.")
```

W8A9
Yêu cầu
Lớp Student(name) với:
add_score(subject, score)
avg() trả về điểm trung bình.
rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8,
"Average" nếu 5–<6.5, ngược lại "Poor".
Input
Dòng 1: name
Dòng 2: n
n dòng tiếp: subject score (score là số thực 0–10)
Output
name avg rank (avg 2 chữ số)

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []
    def add_score(self, subject, score):
        self.scores.append((score))
    def get_average_score(self):
        if not self.scores:
            return 0
        return sum(self.scores) / len(self.scores)
    def get_ranking(self):
        avg = self.get_average_score()
        if avg >= 8:
            return "Excellent"
        elif 6.5 <= avg < 8:
            return "Good" 
        elif 5 <= avg < 6.5:
            return "Average" 
        else:
            return "Poor"
if __name__ == "__main__":
    name = input("Nhập tên sinh viên: ") 
    student = Student(name)
    while True:
        n = input("Nhập số lượng môn học: ") 
        n = int(n)
        for _ in range(n):
            line = input().split()
            subject = " ".join(line[:-1])
            score = float(line[-1])
            student.add_score(subject, score)
        avg_score = student.get_average_score()
        ranking = student.get_ranking()
        print(f"Điểm trung binh của {student.name} là: {avg_score:.2f}, xếp loại: {ranking}") 
        break
```

\(double\) biểu diễn phần thực và phần ảo của một số phức.
Viết hàm khởi tạo ComplexNumber(double initRealPart, double
initImaginaryPart) để khởi tạo các giá trị phần thực và phần ảo

multiply(complexNumber other) trả về tích của số phức hiện tại
và một số phức khác.
Đầu vào
Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực
cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo

Đầu ra
In ra màn hình hai số biểu diễn phần thực và phần ảo của tích
hai số phức.


```python
class ComplexNumber:
    def __init__(self, real = 0, imag = 0):
        self.real = real
        self.imag = imag
    def multiply(self, other):
        real_part = self.real * other.real - self.imag * other.imag
        imag_part = self.real * other.imag + self.imag * other.real
        return ComplexNumber(real_part, imag_part)
if __name__ == "__main__":
    print("Nhập số phức thứ nhất: ") 
    real1, imag1 = map(float, input().split())
    print("Nhập số phức thứ hai: ") 
    real2, imag2 = map(float, input().split())
    complex1 = ComplexNumber(real1, imag1)
    complex2 = ComplexNumber(real2, imag2)
    result = complex1.multiply(complex2)
    print(f"Kết quả phép nhân của hai số phức là: {result.real} + {result.imag}i") 
```

Khai báo cấu trúc dữ liệu tam giác \(Triangle\) chứa \(3\) biến
\(egde1,\,edge2,\,edge3\) kiểu \(int\) là độ dài \(3\) cạnh của
tam giác đó.
Viết hàm khởi tạo Triangle(int initEdge1,int initEdge2, int
initEdge3) để khởi tạo độ dài các cạnh của tam giác và hàm
double getArea() trả về diện tích của tam giác đó (nếu tam giác
không hợp lệ, trả về \(-1\)).
Hoàn thiện chương trình dưới đây nhập vào \(3\) cạnh của một
tam giác và in ra diện tích của tam giác đó.
Đầu vào
Một dòng duy nhất từ bàn phím chứa \(3\) số nguyên là độ dài
\(3\) cạnh của tam giác
Đầu ra
In ra màn hình diện tích tam giác (làm tròn đến\(2\) chữ số thập
phân sau dấu phẩy). In ra \(invalid\) nếu \(3\) số nhập vào
không phải là \(3\) cạnh của một tam giác 


```python
class Triangle:
    def __init__(self, A, B, C):
        self.A = A
        self.B = B
        self.C = C
    def get_angle(self):
        if self.A + self.B > self.C and self.A + self.C > self.B and self.B + self.C > self.A:
            p = (self.A + self.B + self.C) / 2
            area = (p * (p - self.A) * (p - self.B) * (p - self.C)) ** 0.5
            return area
        else:
            return -1.0
if __name__ == "__main__":
    print("Nhập độ dài 3 cạnh của tam giác: ")
    A, B, C = map(float, input().split())
    triangle = Triangle(A, B, C)
    print(f"Diện tích của tam giác là: {triangle.get_angle()}")
```

- Employee: trả về salary
- Manager: trả về salary + bonus
Input

Dòng 2: name salary [bonus]
Output
In ra tổng thu nhập của đối tượng. 


```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def income(self):
        return self.salary
class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus
    def income(self):
        return self.salary + self.bonus
if __name__ == "__main__":
    print("Nhập tên nhân viên: ") 
    name = input().strip()
    print("Nhập lương cơ bản: ") 
    salary = float(input())
    print("Nhập tiền thưởng (nếu có, nếu không thì nhập 0): ") 
    bonus = float(input())
    if bonus > 0:
        employee = Manager(name, salary, bonus)
        print(f"Tổng thu nhập của nhân viên {employee.name} là: {employee.income()}")
    else:
        employee = Employee(name, salary)
        print(f"Tổng thu nhập của nhân viên {employee.name} là: {employee.income()}") 
```
